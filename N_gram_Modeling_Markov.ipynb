```python
!pip install nltk
import random
import nltk
nltk.download('punkt')
```

Data Preprocessing:
We'll use a sample text for demonstration purposes. Replace this with your own text or load a dataset.

```python
text = "This is a sample text for demonstrating N-gram modeling with Markov Chains. N-gram models are used for text generation."

# Tokenize the text into words
tokens = nltk.word_tokenize(text)
```

N-gram Model Creation:
In this step, we'll create an N-gram model with a specified order (N).

```python
def create_ngram_model(tokens, order=2):
    ngram_model = {}

    for i in range(len(tokens) - order):
        ngram = tuple(tokens[i:i+order])
        next_token = tokens[i+order]

        if ngram in ngram_model:
            ngram_model[ngram].append(next_token)
        else:
            ngram_model[ngram] = [next_token]

    return ngram_model

ngram_order = 2  # You can change this to experiment with different N-gram orders
model = create_ngram_model(tokens, order=ngram_order)
```

Text Generation:
Now, let's generate text using the created N-gram model.

```python
def generate_text(model, seed, length=20):
    current_ngram = tuple(seed.split()[-ngram_order:])
    generated_text = list(current_ngram)

    for _ in range(length):
        if current_ngram in model:
            next_possible_tokens = model[current_ngram]
            next_token = random.choice(next_possible_tokens)
            generated_text.append(next_token)
            current_ngram = tuple(generated_text[-ngram_order:])
        else:
            break

    return ' '.join(generated_text)

seed = "This is a"
generated_text = generate_text(model, seed, length=50)
print(generated_text)
```

Now you have a complete code snippet that demonstrates N-gram modeling using Markov Chains for text generation. You can experiment with different orders of N-grams and try generating text with various starting seeds.
