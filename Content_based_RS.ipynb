```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel
```

```python
from google.colab import drive 
drive.mount('/content/gdrive')

movies = pd.read_csv('gdrive/MyDrive/RS/movies.csv')
ratings = pd.read_csv('gdrive/MyDrive/RS/ratings.csv')
```

```python
#Preprocess the data
movies['genres'] = movies['genres'].apply(lambda x: x.replace('|', ' '))
```

```python
#Create a TF-IDF matrix for movie genres:
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vectorizer.fit_transform(movies['genres'])
```

```python
#Calculate the cosine similarity matrix:

cosine_sim = linear_kernel(tfidf_matrix, tfidf_matrix)
```

```python
#Define a function to get movie recommendations:
def get_recommendations(title, cosine_sim=cosine_sim):
    idx = movies[movies['title'] == title].index[0]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:11]
    movie_indices = [i[0] for i in sim_scores]
    return movies.iloc[movie_indices]
```

```python
#Test the recommender system:
recommendations = get_recommendations('Toy Story (1995)')
print(recommendations)
```
