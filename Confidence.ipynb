```python
import numpy as np
import matplotlib.pyplot as plt
```

For simplicity, we'll assume that each item has a true average rating of 4, and the ratings have a standard deviation of 0.5.

```python

# Parameters
true_mean = 4
std_deviation = 0.5
num_items = 1000  # Number of items

# Simulate ratings for each item
np.random.seed(42)  # For reproducibility
ratings_per_item = 10  # Number of ratings per item
ratings = np.random.normal(true_mean, std_deviation, size=(num_items, ratings_per_item))
```

 calculate the sample mean for each item by averaging its ratings.

```python
sample_means = np.mean(ratings, axis=1)
```

 plot the histogram of the sample means to see how the distribution changes as we collect more ratings.

```python

# Plotting the histogram of sample means
plt.hist(sample_means, bins=20, density=True, alpha=0.7, color='blue', label='Sample Means')
plt.axvline(x=true_mean, color='red', linestyle='--', label='True Mean')
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.title('Distribution of Sample Means')
plt.legend()
plt.show()
```

simulate collecting more ratings for each item and see how the distribution of the sample mean changes.

```python
more_ratings_per_item = 100  # More ratings per item
more_ratings = np.random.normal(true_mean, std_deviation, size=(num_items, more_ratings_per_item))

more_sample_means = np.mean(more_ratings, axis=1)

# Plotting the histogram with more ratings
plt.hist(sample_means, bins=20, density=True, alpha=0.5, color='blue', label='Original Sample Means')
plt.hist(more_sample_means, bins=20, density=True, alpha=0.5, color='green', label='More Ratings Sample Means')
plt.axvline(x=true_mean, color='red', linestyle='--', label='True Mean')
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.title('Distribution of Sample Means')
plt.legend()
plt.show()
```

The blue histogram represents the distribution of sample means with the original number of ratings, and the green histogram represents the distribution of sample means with more ratings. As you can see, the distribution becomes narrower and more concentrated around the true mean (red dashed line) as you collect more ratings.
